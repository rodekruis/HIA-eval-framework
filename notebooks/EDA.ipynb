### General Conclusion #1

After cross-checking the csv file at the rows of data that were indexed here, I can confirm the encoding issue has been solved. I have also noticed that the synthetic generation process produces clean, well-formed queries that don't reflect how real users actually write. A real undocumented migrant messaging the helpline might write something like "i dont have papers and i got sick what do i do", which is messy, short, emotionally loaded, possibly in broken English. The synthetic queries look more like "What are the eligibility criteria for undocumented migrants to access healthcare services in the Netherlands?". They are almost academic. However, this does not present an issue, as the validation set's purpose is simply for metric calibration, i.e., ensuring that the metrics work under ideal conditions. The dataset is also just a playground for me to experiment on.
