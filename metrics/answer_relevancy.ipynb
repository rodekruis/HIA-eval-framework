```python
import os
import pandas as pd
import numpy as np
from rich import print
import ast
from deepeval import evaluate
from deepeval.test_case import LLMTestCase
from deepeval.metrics import AnswerRelevancyMetric
from dotenv import load_dotenv, dotenv_values
from deepeval.models import AzureOpenAIModel
```

```python
goldens = pd.read_csv("../synthetic_data/goldens.csv")
```

```python
endpoint = "https://510-ai-research.openai.azure.com/"
model = "gpt-4.1"
deployment = "gpt-4.1-students"

# loading variables from .env file
load_dotenv(r"C:\Users\dari\Desktop\grad project 2026\HIA-eval-framework\.env")
subscription_key = os.getenv("azure_subscription_key")
api_version = "2024-12-01-preview"

# custom model
custom_model = AzureOpenAIModel(
  model=deployment,
  api_key=subscription_key,
  azure_endpoint=endpoint,
  api_version=api_version,
  deployment_name=deployment
)
```

```python
actual_output = goldens['expected_output'][0]
expected_output = goldens['expected_output'][0]

retrieval_context = ast.literal_eval(goldens['context'][0])

metric = AnswerRelevancyMetric(
    threshold=0.7,
    include_reason=True,
    model = custom_model,
)
test_case = LLMTestCase(
    input="Where can I find info about support orgs and shelters for UM, incl. local options?",
    actual_output=actual_output,
    expected_output=expected_output,
)

# To run metric as a standalone
# metric.measure(test_case)
# print(metric.score, metric.reason)

result = evaluate(test_cases=[test_case], metrics=[metric])
```

```python
metric = AnswerRelevancyMetric(
    threshold=0.7,
    include_reason=True,
    model = custom_model,
)

test_cases = []
for i in range(len(goldens)):
    tc = LLMTestCase(
        input=goldens['input'][i],
        actual_output=goldens['expected_output'][i],
        expected_output=goldens['expected_output'][i],
        retrieval_context=ast.literal_eval(goldens['context'][i]),
        context=ast.literal_eval(goldens['context'][i])
    )
    test_cases.append(tc)

print(f"Built {len(test_cases)} test cases")
print(f"First input: {test_cases[0].input[:80]}...")
```
